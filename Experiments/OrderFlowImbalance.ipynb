```python
import pandas as pd
import numpy as np

# ── constants ────────────────────────────────────────────────────────────────
N_LEVELS = 5
LEVEL_W  = [1.0, 0.8, 0.6, 0.4, 0.2]


# ── step 1: per-snapshot pressure ratio ──────────────────────────────────────
def _snapshot_pressure(snap):
    buy_qtys  = [lvl['quantity'] for lvl in snap['buy']]
    sell_qtys = [lvl['quantity'] for lvl in snap['sell']]

    if buy_qtys[0] == 0 and sell_qtys[0] == 0:
        return None

    wb = sum(w * q for w, q in zip(LEVEL_W, buy_qtys))
    wa = sum(w * q for w, q in zip(LEVEL_W, sell_qtys))

    total = wb + wa
    if total == 0:
        return None

    return wb / total


# ── step 2: aggregate snapshots in a block ────────────────────────────────────
def _compute_block_pressure_raw(depth_list):
    if not depth_list:
        return None

    ratios = [_snapshot_pressure(snap) for snap in depth_list]
    ratios = [r for r in ratios if r is not None]

    if not ratios:
        return None

    tw          = np.linspace(0.5, 1.0, len(ratios))
    pressure_tw = float(np.average(ratios, weights=tw))
    pressure_raw = float(np.mean(ratios))
    valid_snaps  = len(ratios)
    total_snaps  = len(depth_list)

    return {
        'pressure_ratio':   round(pressure_tw,  4),
        'pressure_raw':     round(pressure_raw, 4),
        'valid_snap_count': valid_snaps,
        'total_snap_count': total_snaps,
        'data_quality':     round(valid_snaps / total_snaps, 2),
    }


def _empty_result():
    return {
        'pressure_ratio':     np.nan,
        'pressure_raw':       np.nan,
        'valid_snap_count':   0,
        'total_snap_count':   0,
        'data_quality':       0.0,
        'z_pressure':         np.nan,
        'sentiment':          'NEUTRAL',
        'sentiment_strength': 'NEUTRAL',
    }


# ── step 3: full pipeline ─────────────────────────────────────────────────────
def add_depth_signals(df):
    raw_results = df['depth_list'].apply(_compute_block_pressure_raw)

    # replace None with empty dict so DataFrame builds correctly
    raw_df = pd.DataFrame(
        [r if r is not None else {} for r in raw_results],
        index=df.index
    )

    # fill any missing columns that came from None rows
    for col in ['pressure_ratio', 'pressure_raw', 'valid_snap_count', 'total_snap_count', 'data_quality']:
        if col not in raw_df.columns:
            raw_df[col] = np.nan

    mean = raw_df['pressure_ratio'].mean()
    std  = raw_df['pressure_ratio'].std()

    raw_df['z_pressure'] = (raw_df['pressure_ratio'] - mean) / std

    def classify(z):
        if pd.isna(z):
            return 'NEUTRAL', 'NEUTRAL'
        if z >= 1.5:
            return 'BUY',  'STRONG'
        elif z >= 0.75:
            return 'BUY',  'WEAK'
        elif z <= -1.5:
            return 'SELL', 'STRONG'
        elif z <= -0.75:
            return 'SELL', 'WEAK'
        else:
            return 'NEUTRAL', 'NEUTRAL'

    raw_df[['sentiment', 'sentiment_strength']] = raw_df['z_pressure'].apply(
        lambda z: pd.Series(classify(z))
    )
    return pd.concat([df, raw_df], axis=1)


# ── run it ────────────────────────────────────────────────────────────────────
clubbed_df = add_depth_signals(clubbed_df)

print(clubbed_df['sentiment'].value_counts())
print(clubbed_df['sentiment_strength'].value_counts())
```
