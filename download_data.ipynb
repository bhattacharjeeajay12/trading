```python
from dotenv import load_dotenv
import sys
import os
from kiteconnect import KiteConnect
from datetime import datetime, timedelta
import time
import pandas as pd
from pathlib import Path
sys.path.append(os.path.abspath(""))
load_dotenv()
```

```python
# Get Access token
with open(os.path.join("assets", "loginInfo", "access_token.txt"), 'r', encoding='utf-8') as file:
    access_token = file.read()
```

```python
# PARAMETERS

instrument_token = 256265  # NIFTY
file_name = "nifty_1min_full.json"
target_date = '2026-04-16' #should be N


start_date = datetime(2020, 1, 1)
date_obj = datetime.strptime(target_date, '%Y-%m-%d')
formatted_date = date_obj.strftime("%d%b%Y").upper()
next_day_obj = date_obj + timedelta(days=1)
end_date = datetime(next_day_obj.year, next_day_obj.month, next_day_obj.day) # If N is your target date, then put N+1
delta = timedelta(days=60)
```

```python
kite = KiteConnect(api_key=os.getenv('KITE_API_KEY'),)
kite.set_access_token(access_token)

while start_date < end_date:
    chunk_end = min(start_date + delta, end_date)

    print(f"Fetching: {start_date} → {chunk_end}")

    for attempt in range(3):
        try:
            data = kite.historical_data(
                instrument_token=instrument_token,
                from_date=start_date,
                to_date=chunk_end,
                interval="minute"
            )
            break
        except Exception as e:
            print("Retrying...", e)
            time.sleep(1)

    df = pd.DataFrame(data)
    # path = Path("D:/Study/Programs/trading/assets/historical") / file_name
    # df.to_csv(
    #     path,
    #     mode='a',
    #     header=not path.exists(),
    #     index=False
    # )

    path = Path("assets/historical") / f"{file_name}"

    df.to_json(
        path,
        orient="records",
        lines=True,
        mode="a")

    time.sleep(0.4)
    start_date = chunk_end + timedelta(seconds=1)

print("Download complete")
```

```python
import pandas as pd
import json

def parse_date(x):
    try:
        if isinstance(x, dict):
            return x['date']
        if isinstance(x, str):
            return json.loads(x.rstrip('}') + '}')['date']
        return x
    except:
        return None

nifty_all['date'] = (
    pd.to_datetime(nifty_all['date'].map(parse_date), unit='ms', errors='coerce', utc=True)
    .dt.tz_convert('Asia/Kolkata')
)

day_df = nifty_all.loc[
    nifty_all['date'].dt.date == pd.to_datetime(date_obj).date()
]
```
