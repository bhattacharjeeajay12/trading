```python
import json
import re

log_file = Path(rf"D:\Study\Programs\trading\assets\logs\{date_}\streamer.log")
securities = set()
with open(log_file, "r") as f:
    for line in f:
        if "WINDOW_SNAPSHOT" in line:
            # extract JSON part
            match = re.search(r'WINDOW_SNAPSHOT:\s*(\{.*\})', line)
            if not match:
                continue

            data = json.loads(match.group(1))

            for key, value in data.items():
                if key in ["time", "nifty", "atm"]:
                    continue
                securities.add(value)

# result
unique_securities = sorted(securities)
unique_securities = ["NSE:"+security for security in unique_securities]
print(unique_securities)
print(len(unique_securities))
```

```python
HEADERS = {
    "Authorization": auth,
    "Accept": "application/json"
}
```

```python
URL = "https://api-t1.fyers.in/indus/data/v1/timesales"

for symbol in unique_securities:

    PARAMS = {
        "symbol": symbol
    }
    # print(symbol)
    response = requests.get(URL, headers=HEADERS, params=PARAMS)
    data_ = response.json()['d']['data']
    path_ = Path(rf"D:\Study\Programs\trading\assets\logs\{date_}\tns")
    path_.mkdir(parents=True, exist_ok=True)
    data = pd.DataFrame(data_, columns=['time', 'ltp', 'quantity'])
    file_name = symbol.split(":")[-1]
    print(os.path.join(path_, file_name+".xlsx"))
    data.to_excel(os.path.join(path_, file_name +".xlsx"), index=False)
```
