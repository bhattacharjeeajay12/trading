```python
import pandas as pd
import os
from pathlib import Path
```

```python
folder = "30APR2026"
path_ = Path(rf"D:\Study\Programs\trading\assets\logs\{folder}\option_chain_{folder}.xlsx")
output_path = Path(f"assets/MostActiveOptions.csv")

cols = [
    'ATM-3_CE', 'ATM-3_PE', 'ATM-2_CE', 'ATM-2_PE', 'ATM-1_CE', 'ATM-1_PE',
    'ATM_CE', 'ATM_PE', 'ATM+1_CE', 'ATM+1_PE', 'ATM+2_CE', 'ATM+2_PE',
    'ATM+3_CE', 'ATM+3_PE'
]
```

```python
import pandas as pd
import os

def append_two_rows_per_date(df, columns_to_analyze, date_string, output_path):

    # --- get only existing columns ---
    cols = [c for c in columns_to_analyze if c in df.columns]
    if not cols:
        print("No valid columns found")
        return

    # --- flatten + clean ---
    s = pd.Series(df[cols].values.flatten()).dropna()

    # --- counts ---
    vc = s.value_counts()

    # --- build 2 rows ---
    row1 = [date_string] + vc.index.tolist()
    row2 = ["COUNT"] + vc.values.tolist()

    new_rows = pd.DataFrame([row1, row2])

    # --- read existing file (NO headers) ---
    if os.path.exists(output_path):
        old = pd.read_csv(output_path, header=None)
        final = pd.concat([new_rows, old], ignore_index=True)
    else:
        final = new_rows

    # --- write back (NO headers) ---
    final.to_csv(output_path, index=False, header=False)

    print("Done.")
```

```python
df = pd.read_excel(path_)
append_two_rows_per_date(df, cols, folder, output_path)
```
